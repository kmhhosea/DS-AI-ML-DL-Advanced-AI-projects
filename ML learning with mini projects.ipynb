# API to dataframe

```python
import pandas as pd
import requests
```

```python
response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page=1')
```

```python
response.json()
```

```python
response.json()['results']
```

```python
df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
```

```python
df.head()
```

```python
df.shape
```

```python
frames = [df]

for i in range(1, 429):
    response = requests.get(
        'https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={}'.format(
            i)
    )
    temp_df = pd.DataFrame(response.json()['results'])[
        ['id', 'title', 'overview', 'release_date', 'popularity', 'vote_average', 'vote_count']
    ]
    frames.append(temp_df)
df = pd.concat(frames, ignore_index=True)
```

```python
df
```

```python
df.shape
```

```python
df.to_csv('data/movies.csv')
```

Opening a csv file from an URL
